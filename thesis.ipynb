```python
# Satellite Data / Dataset

import numpy as np

# segnale con rumore
time = np.linspace(0, 10, 1000)
signal = np.sin(time)

# aggiungo rumore
noise = np.random.normal(0, 0.5, 1000)
data = signal + noise

# aggiungo un "evento"
data[400:410] += 5

print(data)
```

```python
# Signal Processing

import matplotlib.pyplot as plt

plt.plot(data)
plt.title("Signal with Noise and Event")
plt.show()
```

```python
# Event Detection (ML)

threshold = 3

events = np.where(data > threshold)[0]

print(events[:10])
```

```python
# Anomaly Detection (AI Agent Layer)

from sklearn.ensemble import IsolationForest

model = IsolationForest(contamination=0.01)
data_reshaped = data.reshape(-1, 1)

model.fit(data_reshaped)
pred = model.predict(data_reshaped)

anomalies = np.where(pred == -1)[0]

print(anomalies[:10])
```

```python
# Database (MongoDB)

events_list = [{"index": int(i), "value": float(data[i])} for i in events]

print(events_list[:5])
```
